El enunciado pide condiciones de Big Bang Nucleosynthesis en tres snapshots:

\[
(t,T_9,\rho) \simeq (50\,\mathrm{s},1.7,2\times10^{-4}\,\mathrm{g\,cm^{-3}}),
\]
\[
(200\,\mathrm{s},1.0,2\times10^{-5}\,\mathrm{g\,cm^{-3}}),
\]
\[
(1000\,\mathrm{s},0.4,2\times10^{-6}\,\mathrm{g\,cm^{-3}}),
\]

con razón inicial \(n/p = 1/7\).

En estos notebooks haremos una aproximación mínima: red nuclear + historia \(T(t)\), \(\rho(t)\) impuesta.

```python
# Condiciones termodinámicas del ejercicio
snapshots = pd.DataFrame({
    "label": ["snapshot_50s", "snapshot_200s", "snapshot_1000s"],
    "t_s": [50.0, 200.0, 1000.0],
    "T9": [1.7, 1.0, 0.4],
    "rho_g_cm3": [2.0e-4, 2.0e-5, 2.0e-6],
})
snapshots["T_K"] = snapshots["T9"] * 1.0e9
snapshots
```

```python
# Condición inicial n/p = 1/7
# Como n y p tienen A = 1, las fracciones molares Y coinciden con las fracciones por barión.
n_over_p = 1.0 / 7.0
Yp0 = 1.0 / (1.0 + n_over_p)
Yn0 = n_over_p / (1.0 + n_over_p)

print(f"Yp0 = {Yp0:.8f}")
print(f"Yn0 = {Yn0:.8f}")
print(f"Yn0/Yp0 = {Yn0/Yp0:.8f}")
print(f"Yn0 + Yp0 = {Yn0 + Yp0:.8f}")
```

```python
# Guardamos condiciones para que los otros notebooks sean reproducibles.
snapshots.to_csv("bbn_snapshots.csv", index=False)

with open("bbn_initial_np.txt", "w") as f:
    f.write(f"n_over_p = {n_over_p:.16e}\n")
    f.write(f"Yp0 = {Yp0:.16e}\n")
    f.write(f"Yn0 = {Yn0:.16e}\n")

print("Escritos: bbn_snapshots.csv, bbn_initial_np.txt")
```
